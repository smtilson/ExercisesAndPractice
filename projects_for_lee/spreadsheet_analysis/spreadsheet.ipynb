```python
import pandas as pd
import csv
```

We will be workind to modify a spreadsheet for Lee.

```python
df_original = pd.read_excel('lees_spreadsheet.xlsx', sheet_name='Sheet1')
df_original.info()
df_original.columns
```

The result Lee wanted was to have only the unique rows show up.

```python
cols = df_original.columns
print(cols)
cols1 = [col for col in cols if 'Unnamed' not in col]
print(cols1)
df_w = df_original[cols1]
df_w.head()
```

```python
df_wo = df_w[cols[:2]]
df_wo.head()
```

We are going to accomplish this by forming the set of tuples generated by the rows.

```python
unique_rows_wo = set()
for row in df_wo.itertuples():
    row_tuple = (row[1], row[2])
    if row_tuple not in unique_rows_wo:
        unique_rows_wo.add(row_tuple)

print(len(unique_rows_wo))
print(len(df_wo))
# Save the unique rows to a new Excel file
df_unique = pd.DataFrame(list(unique_rows_wo), columns=[list(cols[:2])])
df_unique.to_excel('unique_rows_wo.xlsx', index=False)
```

```python
unique_rows_w = set()
for row in df_w.itertuples():
    row_tuple = (row[1], row[2])
    if row_tuple not in unique_rows_w:
        unique_rows_w.add(row_tuple)

print(len(unique_rows_w))
print(len(df_w))
# Save the unique rows to a new Excel file
df_unique = pd.DataFrame(list(unique_rows_w), columns=[list(cols1)])
df_unique.to_excel('unique_rows_w.xlsx', index=False)
```
