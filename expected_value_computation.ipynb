# Expected Value of Agents in Initial 5-Card Draw

Computing the expected number of agents when drawing 5 cards from a 36-card deck containing 6 agents.
This follows a hypergeometric distribution.

```python
from math import comb

# Parameters
N = 36  # Total cards in deck
K = 6   # Number of agents in deck
n = 5   # Cards drawn

# Expected value using hypergeometric distribution formula
# E[X] = n * K / N
expected_value_formula = n * K / N

print('Hypergeometric Distribution - Expected Value Calculation')
print('='*60)
print(f'Deck size (N): {N}')
print(f'Number of agents in deck (K): {K}')
print(f'Cards drawn (n): {n}')
print(f'Number of non-agents: {N - K}')
print()
print('Using the hypergeometric expected value formula:')
print(f'E[X] = n * K / N')
print(f'E[X] = {n} * {K} / {N}')
print(f'E[X] = {n * K} / {N}')
print(f'E[X] = {expected_value_formula}')
print(f'E[X] ≈ {expected_value_formula:.6f}')
print()
```

```python
# Verify by computing all possible outcomes
print('Verification - Computing all possible outcomes:')
print('-'*60)
total_expected = 0

for k in range(min(n, K) + 1):  # k = 0, 1, 2, 3, 4, 5, 6
    # But k can't exceed n (5) or K (6), so k = 0 to 5
    if k > n:
        continue
    # P(X = k) = C(K, k) * C(N-K, n-k) / C(N, n)
    prob = (comb(K, k) * comb(N - K, n - k)) / comb(N, n)
    contribution = k * prob
    total_expected += contribution
    print(f'P(X = {k}) = C({K},{k}) * C({N-K},{n-k}) / C({N},{n}) = {prob:.6f}, contribution to E[X]: {contribution:.6f}')

print()
print(f'Total Expected Value: {total_expected:.6f}')
print()
print(f'Final Answer: E[X] = {expected_value_formula} or {expected_value_formula:.6f} agents')
print(f'As a fraction: 30/36 = 5/6 agents')
```
